```python
"""
Rename images inside subfolders that match the pattern: number-number (e.g., 123-456).
For each image, OCR the text, extract:
  - Firstname: ...
  - Lastname:  ...
Then rename the image to:
  Firstname Lastname.<original_extension>
Requirements:
  pip install pytesseract pillow opencv-python
  Install Tesseract OCR:
    - Windows: https://github.com/UB-Mannheim/tesseract/wiki
  Then set TESSERACT_CMD below if needed.
"""

from __future__ import annotations
import os
import re
import sys
import shutil
from pathlib import Path
from typing import Optional, Tuple

import cv2
import pytesseract

# --- CONFIG ---
ROOT_DIR = Path(r"C:\Users\sanaz\Downloads\Final_Craft-20260222T225357Z-1-001\Final_Craft")

# Folder name must look like "123-456"
FOLDER_RE = re.compile(r"^\d+-\d+$")

# Image extensions to process
IMG_EXTS = {".jpg", ".jpeg", ".png", ".webp", ".tif", ".tiff", ".bmp"}

# --- OCR + PARSING HELPERS ---

def preprocess_for_ocr(img_bgr):
    """Improve OCR readability (depends on the images)."""
    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    # denoise
    gray = cv2.bilateralFilter(gray, 9, 75, 75)
    # increase contrast (adaptive)
    thr = cv2.adaptiveThreshold(
        gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 31, 10
    )
    return thr

def ocr_text(image_path: Path) -> str:
    img = cv2.imread(str(image_path))
    if img is None:
        return ""
    proc = preprocess_for_ocr(img)
    #two OCR passes: processed + original gray
    txt1 = pytesseract.image_to_string(proc, lang="eng")
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    txt2 = pytesseract.image_to_string(gray, lang="eng")
    # Combine
    return (txt1 + "\n" + txt2).strip()

def clean_name(s: str) -> str:
    # Keep letters, spaces, apostrophes, hyphens (for names like O'Neil, Anne-Marie)
    s = s.strip()
    s = re.sub(r"[\r\n\t]+", " ", s)
    s = re.sub(r"\s{2,}", " ", s)
    s = re.sub(r"[^A-Za-zÀ-ÖØ-öø-ÿ'\- ]+", "", s)  # includes accented letters
    return s.strip()

def extract_first_last(text: str) -> Tuple[Optional[str], Optional[str]]:
    """
    Extracts Firstname and Lastname from OCR text.
    Handles variations like:
      Firstname: John
      First name : John
      Lastname: Doe
      Last name: Doe
    """
    t = text
    # Normalize common OCR weirdness
    t = t.replace("ﬁ", "fi").replace("ﬂ", "fl")
    t = re.sub(r"[ ]{2,}", " ", t)
    first_patterns = [
        r"First\s*name\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"Firstname\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"First\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
    ]
    last_patterns = [
        r"Last\s*name\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"Lastname\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"Last\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"Surname\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
        r"Family\s*name\s*[:\-]\s*([A-Za-zÀ-ÖØ-öø-ÿ'\- ]{1,50})",
    ]
    first = None
    last = None
    for p in first_patterns:
        m = re.search(p, t, flags=re.IGNORECASE)
        if m:
            first = clean_name(m.group(1))
            break
    for p in last_patterns:
        m = re.search(p, t, flags=re.IGNORECASE)
        if m:
            last = clean_name(m.group(1))
            break
    # Sometimes OCR captures extra words after name; take first 1-3 tokens
    def trim_tokens(name: Optional[str]) -> Optional[str]:
        if not name:
            return None
        toks = name.split()
        toks = toks[:3]
        return " ".join(toks).strip()
    return trim_tokens(first), trim_tokens(last)

def safe_target_name(folder: Path, first: str, last: str, ext: str, src: Path) -> Path:
    base = f"{first} {last}".strip()
    base = re.sub(r"\s{2,}", " ", base)
    # Windows filename safety
    base = re.sub(r'[<>:"/\\|?*]+', "", base).strip().rstrip(".")
    if not base:
        base = src.stem
    candidate = folder / f"{base}{ext.lower()}"
    if not candidate.exists():
        return candidate
    # If already exists, add a counter
    for i in range(2, 9999):
        candidate = folder / f"{base} ({i}){ext.lower()}"
        if not candidate.exists():
            return candidate
    # Worst case fallback
    return folder / f"{base} ({os.getpid()}){ext.lower()}"

def main():
    if not ROOT_DIR.exists():
        print(f"ROOT_DIR not found: {ROOT_DIR}")
        sys.exit(1)
    renamed = 0
    skipped = 0
    failed = 0
    for sub in ROOT_DIR.iterdir():
        if not sub.is_dir():
            continue
        if not FOLDER_RE.match(sub.name):
            continue
        for img_path in sub.rglob("*"):
            if not img_path.is_file():
                continue
            if img_path.suffix.lower() not in IMG_EXTS:
                continue
            text = ocr_text(img_path)
            first, last = extract_first_last(text)

            if not first or not last:
                print(f"[SKIP] Could not find Firstname/Lastname in: {img_path}")
                skipped += 1
                continue
            target = safe_target_name(img_path.parent, first, last, img_path.suffix, img_path)
            # If the filename is already correct, skip
            if img_path.resolve() == target.resolve():
                skipped += 1
                continue
            try:
                img_path.rename(target)
                print(f"[OK] {img_path.name} -> {target.name}")
                renamed += 1
            except Exception as e:
                print(f"[FAIL] {img_path} : {e}")
                failed += 1
    print("\n--- Summary ---")
    print(f"Renamed: {renamed}")
    print(f"Skipped: {skipped}")
    print(f"Failed : {failed}")

if __name__ == "__main__":
    main()
```

```python
import shutil
from pathlib import Path
def copy_with_increment(src, dest_folder):
    dest_folder = Path(dest_folder)
    dest_folder.mkdir(parents=True, exist_ok=True)
    src_path = Path(src)
    dest_path = dest_folder / src_path.name
    if dest_path.exists():
        stem = dest_path.stem
        suffix = dest_path.suffix
        counter = 2
        while (dest_folder / f"{stem}_{counter}{suffix}").exists():
            counter += 1
        dest_path = dest_folder / f"{stem}_{counter}{suffix}"
    shutil.copy(src_path, dest_path)
    print(f"Copied to: {dest_path}")

```

```python
from __future__ import annotations

import os
import re
import time
import json
import requests
import pandas as pd
from pathlib import Path
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

# =========================
# Config
# =========================
URL = "https://iranhrs.org/%D8%AC%D8%A7%D9%86%D8%A8%D8%A7%D8%AE%D8%AA%DA%AF%D8%A7%D9%86-%D8%A7%D8%B9%D8%AA%D8%B1%D8%A7%D8%B6%D8%A7%D8%AA-%D8%B3%D8%B1%D8%A7%D8%B3%D8%B1%DB%8C-%D8%A2%D8%A8%D8%A7%D9%86-%DB%B9%DB%B8-%DA%A9%D8%A7/"
OUT_DIR = Path("iranhrs_aban98")
IMG_DIR = OUT_DIR / "photos"
CSV_PATH = OUT_DIR / "victims.csv"
HTML_PATH = OUT_DIR / "page.html"

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/145.0.0.0 Safari/537.36"
    ),
    "Accept-Language": "fa,en;q=0.9",
}

TIMEOUT = 30
SLEEP_BETWEEN_DOWNLOADS = 0.5


# =========================
# Helpers
# =========================
def clean_text(text: str) -> str:
    if not text:
        return ""
    text = text.replace("\u200c", " ").replace("\xa0", " ")
    text = re.sub(r"\s+", " ", text).strip()
    return text


def sanitize_filename(name: str, max_len: int = 140) -> str:
    name = clean_text(name)
    # Remove characters invalid on Windows/macOS/Linux
    name = re.sub(r'[<>:"/\\|?*]', "", name)
    name = name.strip(". ")
    if not name:
        name = "unknown"
    return name[:max_len]


def get_extension_from_url(url: str, default: str = ".jpg") -> str:
    path = urlparse(url).path.lower()
    _, ext = os.path.splitext(path)
    if ext in {".jpg", ".jpeg", ".png", ".webp"}:
        return ext
    return default


def is_probable_person_name(text: str) -> bool:
    """
    Heuristic for Persian full names.
    Keeps it conservative to reduce false matches.
    """
    text = clean_text(text)

    if not text:
        return False

    # Reject obvious headings / long sentences
    if len(text) > 60:
        return False
    if any(x in text for x in ["استان", "جانباختگان", "اعتراضات", "تصاویر", "لیست", "اسامی", "این لیست", "تکمیل"]):
        return False
    if "–" in text or ":" in text:
        return False

    # Persian letters + spaces + نیم‌فاصله-like variants
    if not re.fullmatch(r"[آ-یئؤإأةك‌ٔ\s\-()]+", text):
        return False

    word_count = len(text.split())
    return 2 <= word_count <= 5


def fetch_html(url: str) -> str:
    r = requests.get(url, headers=HEADERS, timeout=TIMEOUT)
    r.raise_for_status()
    return r.text


def find_main_article(soup: BeautifulSoup):
    """
    Try multiple common content containers used by WordPress/news themes.
    """
    selectors = [
        "article",
        "div.entry-content",
        "div.tdb-block-inner",
        "div.post-content",
        "div.td-post-content",
        "main",
    ]
    for sel in selectors:
        node = soup.select_one(sel)
        if node:
            return node
    return soup


def get_best_image_url(img_tag) -> str | None:
    """
    Prefer high-res versions if available.
    """
    # Common WordPress attributes
    for attr in ["data-lazy-src", "data-src", "src"]:
        val = img_tag.get(attr)
        if val and val.strip():
            return val.strip()

    srcset = img_tag.get("srcset")
    if srcset:
        candidates = []
        for part in srcset.split(","):
            bits = part.strip().split()
            if not bits:
                continue
            url = bits[0]
            size = 0
            if len(bits) > 1 and bits[1].endswith("w"):
                try:
                    size = int(bits[1][:-1])
                except ValueError:
                    size = 0
            candidates.append((size, url))
        if candidates:
            candidates.sort(reverse=True)
            return candidates[0][1]

    return None


def next_probable_name_after(img_tag) -> str:
    """
    Walk forward from the image in document order until we hit a probable Persian person name.
    """
    # Search nearby text-bearing siblings/descendants
    for elem in img_tag.find_all_next(limit=20):
        if elem.name in {"img", "script", "style"}:
            continue

        text = clean_text(elem.get_text(" ", strip=True))
        if is_probable_person_name(text):
            return text

    return ""


def nearest_province_before(tag) -> str:
    """
    Look backward for a nearby heading mentioning استان ...
    """
    for elem in tag.find_all_previous(limit=50):
        if elem.name in {"h2", "h3", "h4", "p", "strong"}:
            text = clean_text(elem.get_text(" ", strip=True))
            m = re.search(r"استان\s+([آ-یئؤإأةك‌ٔ\s\-]+)", text)
            if m:
                return clean_text(m.group(1))
    return ""


def download_file(url: str, out_path: Path):
    out_path.parent.mkdir(parents=True, exist_ok=True)
    r = requests.get(url, headers=HEADERS, timeout=TIMEOUT, stream=True)
    r.raise_for_status()
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)


# =========================
# Main scrape
# =========================
def scrape():
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    IMG_DIR.mkdir(parents=True, exist_ok=True)

    html = fetch_html(URL)
    HTML_PATH.write_text(html, encoding="utf-8")

    soup = BeautifulSoup(html, "lxml")
    article = find_main_article(soup)

    rows = []
    seen = set()

    images = article.find_all("img")
    print(f"Found {len(images)} <img> tags")

    for i, img in enumerate(images, start=1):
        img_url = get_best_image_url(img)
        if not img_url:
            continue

        img_url = urljoin(URL, img_url)

        # Skip obvious site chrome / logos / tiny icons
        alt_text = clean_text(img.get("alt", ""))
        width = img.get("width")
        height = img.get("height")

        if any(x in img_url.lower() for x in ["logo", "icon", "avatar", "banner"]):
            continue

        # The page uses many repeated generic alts like "جانباختگان اعتراضات سراسری"
        # so we do not trust alt alone for person name extraction.
        person_name_fa = next_probable_name_after(img)
        province = nearest_province_before(img)

        # Skip if we couldn't match a likely name
        if not person_name_fa:
            continue

        key = (person_name_fa, img_url)
        if key in seen:
            continue
        seen.add(key)

        ext = get_extension_from_url(img_url, default=".jpg")
        filename = f"{sanitize_filename(person_name_fa)}{ext}"
        filepath = IMG_DIR / filename

        try:
            if not filepath.exists():
                download_file(img_url, filepath)
                time.sleep(SLEEP_BETWEEN_DOWNLOADS)

            rows.append(
                {
                    "Name in Farsi": person_name_fa,
                    "Province": province,
                    "Image URL": img_url,
                    "Image Filename": filename,
                    "Source URL": URL,
                    "Img Alt": alt_text,
                    "Width": width,
                    "Height": height,
                }
            )
            print(f"[{i}] Saved: {filename}")

        except Exception as e:
            print(f"[{i}] Failed for {person_name_fa}: {e}")

    df = pd.DataFrame(rows).drop_duplicates()
    df.to_csv(CSV_PATH, index=False, encoding="utf-8-sig")

    print(f"\nDone.")
    print(f"Rows: {len(df)}")
    print(f"CSV: {CSV_PATH}")
    print(f"Photos: {IMG_DIR}")


if __name__ == "__main__":
    scrape()
```

```python
from pathlib import Path
import cv2
import numpy as np
import pandas as pd

PHOTOS_DIR = Path(r"C:\Users\sanaz\OneDrive\Documents\ThePriceofLibertyMAP\iranhrs_aban98\photos")
DEBUG_DIR = PHOTOS_DIR.parent / "debug_text_crops_only"
DEBUG_DIR.mkdir(parents=True, exist_ok=True)

def read_image_unicode(path: Path):
    try:
        data = np.fromfile(str(path), dtype=np.uint8)
        if data.size == 0:
            return None
        return cv2.imdecode(data, cv2.IMREAD_COLOR)
    except Exception:
        return None

def save_image_unicode(path: Path, img: np.ndarray) -> bool:
    """
    Unicode-safe image save for Windows paths.
    """
    try:
        ext = path.suffix if path.suffix else ".png"
        ok, buf = cv2.imencode(ext, img)
        if not ok:
            return False
        buf.tofile(str(path))
        return True
    except Exception:
        return False

def crop_text_region(img_bgr):
    h, w = img_bgr.shape[:2]
    x1 = int(w * 0.02)
    x2 = int(w * 0.46)
    y1 = int(h * 0.60)
    y2 = int(h * 0.98)
    return img_bgr[y1:y2, x1:x2]

def preprocess_crop(roi_bgr):
    gray = cv2.cvtColor(roi_bgr, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, None, fx=5, fy=5, interpolation=cv2.INTER_CUBIC)
    gray = cv2.GaussianBlur(gray, (3, 3), 0)
    _, th = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    th = 255 - th
    return th

rows = []

image_paths = sorted(
    [p for p in PHOTOS_DIR.iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png", ".webp"}]
)

for img_path in image_paths:   # test first 20
    img = read_image_unicode(img_path)

    if img is None:
        rows.append({
            "Original Filename": img_path.name,
            "Saved Filename": "",
            "Status": "FAILED_READ"
        })
        continue

    roi = crop_text_region(img)
    proc = preprocess_crop(roi)

    # keep EXACT same original file name
    out_path = DEBUG_DIR / img_path.name
    saved = save_image_unicode(out_path, proc)

    rows.append({
        "Original Filename": img_path.name,
        "Saved Filename": out_path.name if saved else "",
        "Status": "CROP_SAVED" if saved else "FAILED_SAVE"
    })

df = pd.DataFrame(rows)
print(df)
```

```python
from __future__ import annotations

import os
import re
from pathlib import Path

import pandas as pd
import pytesseract
from PIL import Image

# =========================================
# CONFIG
# =========================================
CROPS_DIR = Path(r"C:\Users\sanaz\OneDrive\Documents\ThePriceofLibertyMAP\iranhrs_aban98\debug_text_crops_only")
OUT_CSV = CROPS_DIR.parent / "ocr_crops_tabular_output.csv"

pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
os.environ["TESSDATA_PREFIX"] = r"C:\Program Files\Tesseract-OCR\tessdata"

TESS_CONFIG = "--oem 3 --psm 6 -l fas"
VALID_EXTS = {".jpg", ".jpeg", ".png", ".webp"}


# =========================================
# HELPERS
# =========================================
def normalize_farsi_text(text: str) -> str:
    if not text:
        return ""

    replacements = {
        "\u200c": " ",   # ZWNJ
        "\u200f": "",    # RTL mark
        "\u200e": "",    # LTR mark
        "ك": "ک",
        "ي": "ی",
        "ة": "ه",
        "ۀ": "ه",
        "أ": "ا",
        "إ": "ا",
        "ؤ": "و",
        "‌": " ",
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    lines = []
    for line in text.splitlines():
        line = re.sub(r"\s+", " ", line).strip()
        if line:
            lines.append(line)

    return "\n".join(lines)


def split_lines(text: str) -> list[str]:
    text = normalize_farsi_text(text)
    return [line.strip() for line in text.splitlines() if line.strip()]


def ocr_image(path: Path) -> str:
    img = Image.open(path)
    text = pytesseract.image_to_string(img, config=TESS_CONFIG)
    return normalize_farsi_text(text)


# =========================================
# MAIN
# =========================================
rows = []

image_paths = sorted(
    [p for p in CROPS_DIR.iterdir() if p.is_file() and p.suffix.lower() in VALID_EXTS]
)

print(f"Found {len(image_paths)} crop images.")

for i, img_path in enumerate(image_paths, start=1):
    print(f"[{i}/{len(image_paths)}] OCR: {img_path.name}")

    try:
        raw_text = ocr_image(img_path)
        lines = split_lines(raw_text)

        row = {
            "Image Filename": img_path.name,
            "Line 1": lines[0] if len(lines) > 0 else "",
            "Line 2": lines[1] if len(lines) > 1 else "",
            "Line 3": lines[2] if len(lines) > 2 else "",
            "Line 4": lines[3] if len(lines) > 3 else "",
            "Raw OCR Text": raw_text,
            "OCR Status": "OK",
        }

    except Exception as e:
        row = {
            "Image Filename": img_path.name,
            "Line 1": "",
            "Line 2": "",
            "Line 3": "",
            "Line 4": "",
            "Raw OCR Text": "",
            "OCR Status": f"FAILED: {e}",
        }

    rows.append(row)

df = pd.DataFrame(rows)

# show table in notebook / console
print(df[["Image Filename", "Line 1", "Line 2", "Line 3", "Line 4", "OCR Status"]])
```

```python
from bs4 import BeautifulSoup
from pathlib import Path
import pandas as pd
import shutil
import re

# PATHS — change these
EXPORT_DIR = Path(r"C:\Users\sanaz\Downloads\Telegram Desktop\ChatExport_2026-03-10")
HTML_FILE = EXPORT_DIR / "messages.html"
PHOTOS_DIR = EXPORT_DIR / "photos"

OUTPUT_DIR = EXPORT_DIR / "extracted_results"
OUTPUT_DIR.mkdir(exist_ok=True)

IMG_OUT = OUTPUT_DIR / "photos"
IMG_OUT.mkdir(exist_ok=True)

# read html
with open(HTML_FILE, "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

records = []

messages = soup.find_all("div", class_="message")

for m in messages:

    text = m.get_text("\n")

    match = re.search(r"([۰-۹0-9]+)\.\s*(.+)", text)

    if match:
        name = match.group(2).strip()

        # find next line containing date/location
        lines = text.split("\n")
        date_loc = ""

        for l in lines:
            if "۱۴۰" in l:
                date_loc = l.strip()
                break

        # find image
        img = m.find("img")
        img_file = None

        if img:
            src = img.get("src")
            img_path = EXPORT_DIR / src

            if img_path.exists():
                new_name = f"{name}.jpg"
                new_path = IMG_OUT / new_name
                shutil.copy(img_path, new_path)
                img_file = new_name

        records.append({
            "name_farsi": name,
            "date_location": date_loc,
            "photo": img_file
        })

df = pd.DataFrame(records)

```

```python
# Attempt again by inspecting columns first and cleaning the correct one
import pandas as pd
import re
from pathlib import Path


# Identify the column that contains the date/location (likely second column)
cols = list(df.columns)

def extract_location(text):
    if pd.isna(text):
        return text
    text = str(text)
    text = re.sub(r"[0-9۰-۹]", "", text)
    months = ["فروردین","اردیبهشت","خرداد","تیر","مرداد","شهریور",
              "مهر","آبان","آذر","دی","بهمن","اسفند"]
    for m in months:
        text = text.replace(m, "")
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Apply cleaning to every column except the first (names) to be safe
for c in cols[1:]:
    df[c] = df[c].apply(extract_location)

df.to_excel(OUTPUT_DIR / "records.xlsx", index=False)
df.to_csv(OUTPUT_DIR / "records.csv", index=False)

print("Done.")
print("Results saved to:", OUTPUT_DIR)
```

```python
import pandas as pd
import re

LETTER_MAP = {
"ا":"a","آ":"a","ب":"b","پ":"p","ت":"t","ث":"s","ج":"j","چ":"ch","ح":"h","خ":"kh","د":"d","ذ":"z",
"ر":"r","ز":"z","ژ":"zh","س":"s","ش":"sh","ص":"s","ض":"z","ط":"t","ظ":"z","ع":"","غ":"gh","ف":"f",
"ق":"gh","ک":"k","گ":"g","ل":"l","م":"m","ن":"n","و":"v","ه":"h","ی":"y"
}

VOWEL_RULES = {
"محمد":"Mohammad",
"علی":"Ali",
"رضا":"Reza",
"حسین":"Hossein",
"حسن":"Hasan",
"مهدی":"Mehdi",
"مصطفی":"Mostafa",
"سعید":"Saeed",
"وحید":"Vahid",
"امیر":"Amir",
"امید":"Omid",
"محمود":"Mahmoud",
"مجید":"Majid",
"فرهاد":"Farhad",
"بهروز":"Behrouz",
"بهرام":"Bahram",
"داریوش":"Dariush"
}

def transliterate(name):

    parts = name.split()

    out = []

    for p in parts:

        if p in VOWEL_RULES:
            out.append(VOWEL_RULES[p])
            continue

        latin = ""

        for c in p:
            latin += LETTER_MAP.get(c,c)

        latin = latin.capitalize()

        latin = latin.replace("aa","a").replace("ee","i")

        out.append(latin)

    return " ".join(out)

df["name in english"] = df["name_farsi"].apply(transliterate)

df.to_excel("records_with_corrected_english_names.xlsx", index=False)
df.to_csv("records_with_corrected_english_names.csv", index=False)
```

```python


def parse_mapping(mapping_text: str) -> dict[str, str]:
    mapping = {}
    for line in mapping_text.splitlines():
        line = line.strip()
        if not line:
            continue
        parts = line.split("\t")
        if len(parts) < 2:
            continue
        farsi = parts[0].strip()
        english = parts[1].strip()
        mapping[farsi] = english
    return mapping

def safe_filename(name: str) -> str:
    # Remove Windows-invalid filename characters
    name = re.sub(r'[<>:"/\\|?*]', "", name)
    # normalize whitespace
    name = re.sub(r"\s+", " ", name).strip()
    return name

def find_existing_file(folder: Path, stem: str):
    # Try common image extensions
    exts = [".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG", ".webp", ".WEBP"]
    for ext in exts:
        p = folder / f"{stem}{ext}"
        if p.exists():
            return p
    return None

mapping = parse_mapping(MAPPING_TEXT)

renamed = 0
missing = []
collisions = []

for farsi_name, english_name in mapping.items():
    src = find_existing_file(PHOTOS_DIR, farsi_name)
    if src is None:
        missing.append(farsi_name)
        continue
    english_safe = safe_filename(english_name)
    dst = src.with_name(f"{english_safe}{src.suffix.lower()}")
    if dst.exists() and dst.resolve() != src.resolve():
        collisions.append((farsi_name, english_name, dst.name))
        continue
    src.rename(dst)
    renamed += 1

print(f"Renamed: {renamed}")
print(f"Missing: {len(missing)}")
print(f"Collisions: {len(collisions)}")

if missing:
    print("\nMissing files:")
    for x in missing[:50]:
        print(" -", x)

if collisions:
    print("\nName collisions:")
    for farsi_name, english_name, target in collisions[:50]:
        print(f" - {farsi_name} -> {english_name} (target exists: {target})")


```
